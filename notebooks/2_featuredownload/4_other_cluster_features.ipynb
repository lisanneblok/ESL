```python
import geopandas as gpd
import matplotlib.pyplot as plt

gdf = gpd.read_file("/Users/lb962/Downloads/HydroRIVERS_v10_eu_shp/HydroRIVERS_v10_eu_shp/HydroRIVERS_v10_eu.shp")

# Example: keep main rivers with > 500 km to outlet
main_rivers = gdf[
    (gdf["HYRIV_ID"] == gdf["MAIN_RIV"]) &
    (gdf["DIST_DN_KM"] > 500)
]

fig, ax = plt.subplots(figsize=(10, 12))
main_rivers.plot(ax=ax)
ax.set_title("Major Main Rivers (DIST_DN_KM > 500 km)")
ax.set_xlim(-15, 10)
ax.set_ylim(48, 65)
plt.show()
```
